```python
def crear_tablero():
    import numpy as np
    tablero= np.full((10,10),"[]")
    return tablero
crear_tablero()

tablero_jugador1= crear_tablero()
tablerod_jugador1= crear_tablero()
tablero_rival= crear_tablero()
tablerod_rival= crear_tablero()

def crear_barco(eslora , tablero):
    import random
    while True:
     horizontal = random.choice([True, False])
     validez=True
     barco=[]
     x = random.randrange(0,10)
     y = random.randrange(0,10)
     if horizontal == True and x+eslora<10:
         for i in range(eslora):
             barco.append((x+i,y))
             if tablero[y][x+i]!="[]":         #Hacer con ands
                 validez=False
                 break
             if tablero[y+1][x+i]!="[]":
                 validez=False
                 break
             if tablero[y-1][x+i]!="[]":
                 validez=False
                 break
             if tablero[y][x+i]!="[]":
                 validez=False
                 break
         if validez:
             return barco
     elif horizontal == False and y+eslora<10:
         for i in range(eslora):
             barco.append((x,y+i))
             if tablero[y+i][x]!="[]":
                 validez=False
                 break
         if validez:
             return barco
         
def meter_barco(barco,tablero):
    for i in barco:
        tablero[i[1]][i[0]]="O"
    return tablero

barco = crear_barco(4,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(3,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(3,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)

def mostrar_tablero(tablero):
    for fila in tablero:
        # Alineamos cada elemento para que ocupe exactamente 4 espacios de ancho
        fila_formateada = "".join([str(celda).ljust(4) for celda in fila])
        print(fila_formateada)
    print()

mostrar_tablero(tablero_jugador1)
    
```
