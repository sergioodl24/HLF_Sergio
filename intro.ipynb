```python
import numpy as np
def crear_tablero():
    import numpy as np
    tablero= np.full((10,10),"[ ]")
    return tablero
crear_tablero()

tablero_jugador1= crear_tablero()
tablerod_jugador1= crear_tablero()
tablero_rival= crear_tablero()
tablerod_rival= crear_tablero()

def crear_barco(eslora , tablero):
    import random
    while True:
     horizontal = random.choice([True, False])
     validez=True
     barco=[]
     x = random.randrange(0,10)
     y = random.randrange(0,10)
     if horizontal == True and x+eslora<10:
         for i in range(eslora):
             barco.append((x+i,y))
             if tablero[y][x+i]!="[ ]":         #Hacer con ands
                 validez=False
                 break
             if y < 9 and tablero[y + 1][x + i] != "[ ]":
                 validez = False
                 break
             if y > 0 and tablero[y - 1][x + i] != "[ ]":
                 validez = False
                 break
             if i == 0 and x > 0 and tablero[y][x - 1] != "[ ]":
                 validez = False
                 break
             if i == eslora - 1 and (x + eslora) < 10 and tablero[y][x + eslora] != "[ ]":
                 validez = False
                 break            
         if validez:
             return barco
     elif horizontal == False and y+eslora<10:
         for i in range(eslora):
             barco.append((x,y+i))
             if tablero[y+i][x]!="[]":
                 validez=False
                 break
             if x < 9 and tablero[y + i][x + 1] != "[ ]":
                 validez = False
                 break
             if x > 0 and tablero[y + i][x - 1] != "[ ]":
                 validez = False
                 break
             if i == 0 and y > 0 and tablero[y - 1][x] != "[ ]":
                 validez = False
                 break
             if i == eslora - 1 and (y + eslora) < 10 and tablero[y + eslora][x] != "[ ]":
                 validez = False
                 break
         if validez:
             return barco
         
def meter_barco(barco,tablero):
    for i in barco:
        tablero[i[1]][i[0]]="O"
    return tablero

barco = crear_barco(4,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(3,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(3,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)
barco = crear_barco(2,tablero_jugador1)
tablero_jugador1=meter_barco(barco,tablero_jugador1)

def mostrar_tablero(tablero):
    print("\n      " + "".join(f"{col:<4}" for col in range(1, 11)) + "\n   " + "-" * 43)
    for i, fila in enumerate(tablero, start=1):
        render = [c if c == "[ ]" else c.center(3) for c in fila]
        print(f"{i:2d} | {' '.join(render)} |")
        
    print("   " + "-" * 43 + "\n")



turno_jugador=True

while True:
    existeplayer = "O" in tablero_jugador1
    existerival = True #"O" in tablero_rival
    if existeplayer== False:
        print("has perdido")
        break
    elif existerival== False:
        print("has ganado")
        break
    else:
        if turno_jugador==True:
            disparox=int(input("elige coordenada x (del 1 al 10)"))-1
            disparoy=int(input("elige coordenada y (del 1 al 10)"))-1
            if  disparox < 0 or disparox >= 10 or disparoy < 0 or disparoy >= 10   :
                print ("zona no valida")
                continue
            else:
                if tablero_rival[disparoy][disparox] != "[ ]" and tablero_rival[disparoy][disparox] != "O" :
                    print ("zona no valida")
                    continue
                elif  tablero_rival[disparoy][disparox] == "[ ]":
                    print ("has fallado")
                    tablerod_jugador1 [disparoy][disparox] ="X"
                    tablero_rival [disparoy][disparox] ="X"
                    #limpiar casilla
                    mostrar_tablero(tablerod_jugador1)
                    mostrar_tablero(tablero_jugador1)
                    turno_jugador=False
                    continue
                elif  tablero_rival[disparoy][disparox] == "O":
                    print ("has acertado")
                    tablerod_jugador1 [disparoy][disparox] ="I"
                    tablero_rival [disparoy][disparox] ="I"
                    #limpiar casilla
                    mostrar_tablero(tablerod_jugador1)
                    mostrar_tablero(tablero_jugador1)             
                    continue
        elif turno_jugador==False:
            disparox=numero = np.random.randint(0, 10)
            disparoy=numero = np.random.randint(0, 10)
            if  tablero_jugador1[disparoy][disparox] == "[ ]":
                tablero_jugador1 [disparoy][disparox] ="X"
                #limpiar casilla
                mostrar_tablero(tablerod_jugador1)
                mostrar_tablero(tablero_jugador1)
                turno_jugador=True
                continue
            elif  tablero_rival[disparoy][disparox] == "O":
                tablero_jugador1 [disparoy][disparox] ="I"
                #limpiar casilla
                mostrar_tablero(tablerod_jugador1)
                mostrar_tablero(tablero_jugador1)             
                continue
            elif  tablero_rival[disparoy][disparox] != "[ ]" and tablero_rival[disparoy][disparox] != "O" :
                continue
         


```
